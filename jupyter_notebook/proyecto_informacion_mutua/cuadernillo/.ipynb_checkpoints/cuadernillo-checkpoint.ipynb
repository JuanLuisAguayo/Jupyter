## Motivación

Para tomar decisiones óptimas, es esencial tener una comprensión clara del conocimiento disponible y cómo se relaciona con la decisión en cuestión. 

La información mutua proporciona una medida cuantitativa de la relación entre dos variables, permitiéndonos evaluar la cantidad de información
compartida entre ellas. Al cuantificar esta información compartida, podemos obtener información valiosa sobre las dependencias y patrones que 
existen dentro de nuestros datos.

Desde una perspectiva estadística, la información mutua sirve como una herramienta poderosa para la selección de características y la reducción de 
dimensionalidad. Ayuda a identificar las variables más informativas que contribuyen significativamente a la variable objetivo, lo que nos permite 
centrarnos en aspectos relevantes y descartar información irrelevante o redundante. Esto no sólo simplifica el proceso de toma de decisiones sino
que también mejora su precisión al reducir el ruido y eliminar complejidades innecesarias.

Además, la información mutua desempeña un papel vital en diversos campos, como el aprendizaje automático, la minería de datos y la teoría de la 
información. En el aprendizaje automático, ayuda a construir modelos sólidos al identificar las características más influyentes que contribuyen a 
predicciones precisas. Por ejemplo, en tareas de reconocimiento de imágenes, la información mutua puede ayudar a determinar qué píxeles o regiones
son más informativos para clasificar diferentes objetos o patrones.

En aplicaciones de minería de datos, la información mutua nos permite descubrir relaciones ocultas y dependencias entre variables. Al medir la 
cantidad de información compartida entre dos variables, podemos identificar asociaciones que pueden no ser evidentes mediante un simple análisis de
correlación. Por ejemplo, considere una campaña de marketing en la que queremos comprender qué atributos del cliente son más influyentes a la hora
de predecir el comportamiento de compra. Al calcular la información mutua entre los datos demográficos de los clientes y el historial de compras, 
podemos descubrir información valiosa sobre las preferencias de los clientes y adaptar nuestras estrategias de marketing en consecuencia.

En teoría de la información, la información mutua cuantifica la reducción de la incertidumbre sobre una variable dado el conocimiento de otra 
variable. Mide en qué medida el conocimiento de una variable reduce nuestra incertidumbre sobre el valor de otra variable. Este concepto es 
particularmente útil en sistemas de comunicación donde la transmisión eficiente de información es crucial. Al maximizar la información mutua entre 
la señal transmitida yla señal recibida, podemos garantizar una comunicación confiable y precisa.

## Introducción

La **teoría de la información**, también conocida como **teoría matemática de la comunicación o teoría matemática de la información**, es una 
propuesta teórica presentada por Claude E. Shannon y Warren Weaver a finales de la década de 1940. Esta teoría está relacionada con las leyes  
matemáticas que rigen la transmisión y el procesamiento de la información y se ocupa de la medición de la información y de la representación de la 
misma, así como también de la capacidad de los sistemas de comunicación para transmitir y procesar información [1]. 

"Shannon" entiende la comunicación como un proceso de transferencia de información (...) proporcionó nuevas dimensiones de la construcción de un 
modelo de comunicación humana, aunque era básicamente un modelo físico de las comunicaciones radiales y telefónicas. Desde entonces en el 
vocabulario de la comunicación aparecieron conceptos como entropía, redundancia, bit, capacidad del canal, ruido y feed back.

El modelo propuesto por Shannon es un sistema general de la comunicación que parte de una fuente de información desde la cual, a través de un 
transmisor, se emite una señal, la cual viaja por un canal, pero a lo largo de su viaje puede ser interferida por algún ruido. La señal sale del 
canal, llega a un receptor que decodifica la información convirtiéndola posteriormente en mensaje que pasa a un destinatario.

<img src="Fig1_diagrama_sistema_comunicacion.png" width=600 height=600 />

Con el modelo de la teoría de la información se trata de llegar a determinar la forma más económica, rápida y segura de codificar un mensaje, sin 
que la presencia de algún ruido complique su transmisión. Para esto, el destinatario debe comprender la señal correctamente; el problema es que 
aunque exista un mismo código de por medio, esto no significa que el destinatario va a captar el significado que el emisor le quiso dar al mensaje. 
La codificación puede referirse tanto a la transformación de voz o imagen en señales eléctricas o electromagnéticas, como al cifrado de mensajes 
para asegurar su privacidad. Un concepto fundamental en la teoría de la información es que la cantidad de información contenida en un mensaje es un 
valor matemático bien definido y medible. El término cantidad no se refiere a la cuantía de datos, sino a la probabilidad de que un mensaje, dentro 
de un conjunto de mensajes posibles, sea recibido. 

La teoría de la información es una rama de la teoría de la probabilidad que estudia la información y todo lo relacionado con ella: canales, 
compresión de datos y criptografía, entre otros.

En el ámbito de la teoría de la información la **entropía**, también llamada **entropía de la información** y **entropía de Shannon**, mide la 
incertidumbre de una fuente de información.

La entropía también se puede considerar como la cantidad de información promedio que contienen los símbolos usados. Los símbolos con menor 
probabilidad son los que aportan mayor información; por ejemplo, si se considera como sistema de símbolos a las palabras en un texto, palabras 
frecuentes como "que", "el", "a" aportan poca información, mientras que palabras menos frecuentes como "corren", "niño", "perro" aportan 
más información. Si de un texto dado borramos un "que", seguramente no afectará a la comprensión y se sobreentenderá, no siendo así si borramos 
la palabra "niño" del mismo texto original. Cuando todos los símbolos son igualmente probables (distribución de probabilidad plana), todos 
aportan información relevante y la entropía es máxima.

En teoría de la probabilidad, y en teoría de la información, la **información mutua o transinformación o ganancia de información** de dos variables 
aleatorias es una cantidad que mide la dependencia estadística entre ambas variables. Más precisamente, mide la información o reducción de la 
incertidumbre *(entropía)* de una variable aleatoria, ${\displaystyle X}$, debido al conocimiento del valor de otra variable aleatoria 
${\displaystyle Y}$ [2]. No se limita a variables en números reales ni a relaciones lineales, por lo que la información mutua es más general que el 
coeficiente de correlación de Pearson. 

Esta cantidad fue definida y analizada por Claude Shannon en su trabajo seminal, **A Mathematical Theory of Communication**, aunque él no la llamó 
**información mutua**. El término fue acuñado más tarde por Robert Fano [3].

La información mutua (MI, por sus siglas en inglés) es un concepto fundamental de la teoría de la información que cuantifica la cantidad de 
información que dos variables aleatorias o randómicas proporcionan una sobre la otra. Mide la reducción de incertidumbre de una variable debido al 
conocimiento de la otra \cite{Or2003}.

La información mutua se utiliza ampliamente para cuantificar dependencias entre variables aleatorias y tiene aplicaciones importantes en selección 
de características, clustering, clasificación, teoría de comunicaciones y muchas otras áreas dentro de la informática.\cite{IAA2023}


## Marco Teórico

Cuando hablamos de información en el sentido de Shannon, no nos referimos al significado de un mensaje, sino a cuánto aprendemos al observar un 
resultado.

Imaginemos que ocurre un evento que ya esperábamos casi con total seguridad. Al observarlo, no sentimos sorpresa alguna, pues nuestra idea previa 
del mundo prácticamente no cambia. En este caso, decimos que el evento aporta muy poca información. 

Por el contrario, si ocurre un evento que creíamos poco probable, la situación es distinta. Nos sorprende, nos obliga a actualizar nuestras 
expectativas y, por tanto, decimos que ese resultado contiene mucha información.

La clave está entonces en la probabilidad, esto es,

eventos probables $\rightarrow$ poca información, <br>
eventos improbables $\rightarrow$ mucha información.

Shannon formalizó esta idea asignando a cada resultado un número que mide su sorpresa. 

En la teoría de la información de Shannon, el concepto de información se define de manera estrictamente probabilística, desvinculada de cualquier 
interpretación semántica. La información asociada a un resultado se entiende como la reducción de incertidumbre que dicho resultado produce en un 
observador.

Sea $X$ una variable aleatoria con distribución de probabilidad $p(x)$. La observación de un resultado $x$ aporta información en la medida en que 
dicho resultado era incierto antes de ocurrir. En particular y tal como se mencionó antes, eventos altamente probables producen una reducción 
marginal de la incertidumbre, mientras que eventos poco probables generan una ganancia informativa significativa.

Para cuantificar esta idea, se introduce el contenido de información de Shannon, definido como una función decreciente de la probabilidad del 
resultado. Además, esta función debe satisfacer una propiedad fundamental: la información asociada a resultados independientes debe ser aditiva. 
Estas consideraciones conducen de forma natural a una definición logarítmica.

En consecuencia, el **contenido de información de Shannon asociado a un resultado $x$** se define [4] como: 

$$h(x)=−\log{p(x)},$$

donde la base del logaritmo fija la unidad de medida, esto es,

base 2: bits,<br>
base $e$: nats, <br>
base 10: hartleys o bans.

En palabras simples, el contenido de información de Shannon $h(x)$ cuantifica el grado de sorpresa asociado a observar el resultado $x$, esto 
es,

Si $p(x)$ es grande (evento frecuente), entonces $h(x)$ es pequeño, luego el resultado aporta poca información nueva.<br>
Si $p(x)$ es pequeño (evento infrecuente o raro), entonces $h(x)$ es grande, luego el resultado es altamente informativo.<br>

Por lo tanto, esto refleja directamente la idea de sorpresa: cuanto más raro es un evento, mayor es su contenido informativo.

Las propiedades fundamentales del contenido de información de Shannon son:

1. *No negatividad* $$h(x)\leq0,\qquad \text{ya que}\qquad 0<p_{X}(x)\leq1.$$
2. *Monotonía decreciente*<br>
   A menor probabilidad del evento, mayor contenido de información.
3. *Aditividad para eventos independientes* <br>
   Si $x_1$ y $x_2$ son eventos independientes, $$h(x_1,x_2)=h(x_1)+h(x_2)$$

Veamos el siguiente ejemplo, considérese una variable aleatoria discreta $X$ que puede tomar tres valores:

$$X\in\{a,b,c\}$$

con la siguiente distribución de probabilidad:

$$p(a)=0.5,\qquad p(b)=0.25,\qquad p(c)=0.25.$$ 

Usaremos logaritmos en base 2, por lo que la información se mide en bits.

Por definición, $h(x)=−\log_{2}{p(x)}.$

Si $x=a$, entonces $h(a)=−\log_{2}{0.5}=1\ bit.$ <br>
Si $x=a$, entonces $h(a)=−\log_{2}{0.25}=2\ bit.$ <br>
Si $x=a$, entonces $h(a)=−\log_{2}{0.25}=2\ bit.$ <br>

El contenido de información $h(x)$ no es un promedio ni una propiedad global de la fuente; describe exclusivamente qué ocurre cuando ese resultado 
específico se observa. En este ejemplo, los resultados menos probables presentan un mayor contenido de información que el resultado más probable.

En conclusión, la información no es una propiedad semántica del evento, sino una medida puramente probabilística del carácter inesperado del 
resultado.

Este concepto constituye el bloque elemental sobre el cual se construyen magnitudes fundamentales de la teoría de la información, como la 
entropía, la información mutua y la información condicional, las cuales permiten analizar sistemas de comunicación, procesos de decisión y 
dependencia estadística entre variables aleatorias.

En la teoría de la información, la **entropía** también denominada **entropía de la información o entropía de Shannon**, es una medida de la 
incertidumbre o la aleatoriedad de una fuente de datos, cuantificando la cantidad promedio de información que contiene o libera un mensaje, esto 
es, a mayor entropía, mayor será la imprevisibilidad del mensaje.

Matemáticamente, la entropía de un mensaje $X$, denotado por $H(X)$, es el valor medio ponderado de la cantidad de información de Shannon de los 
diversos estados del mensaje [4], esto es,

$$H(X)\equiv\sum_{x\in{\mathcal{A}_X}}p(x)\log{\frac{1}{p(x)}},$$

con la convención que para $p(x)=0$ tenemos que $0\times\log{1/0}\equiv0$, donde $\lim_{\theta\rightarrow0^{+}}\theta\log{1/\theta}=0$.

Tal como $h(x)$, la entropía se mide en bits y, además, representa una medida de la incertidumbre media acerca de una variable aleatoria y por lo tanto de la cantidad de información.

Algunas de las propiedades de la función entropía son:

1. *No Negativadad*: $H(X)>0$. Esto significa que la incertidumbre asociada a una variable aleatoria no puede ser menor que cero.
2. *Cota Superior*: $H(X)\leq \log{|\mathcal{A}_X|}$, la igualdad se alcanza si y solo $p(x_i)=\frac{1}{|\mathcal{A}_X|}$ para todo $i$.<br>
   Por lo tanto, la máxima incertidumbre ocurre cuando todos los resultados son igualmente probables.<br>
   Notación: las barras verticales $|\cdot|$ tienen dos significados. Si $\mathcal{A}_X$ es un conjunto, $|\mathcal{A}_X|$ denota los números de
   elementos en $\mathcal{A}_X$; si $x$ es un número, entonces $|x|$ es el valor absoluto del número $x$. 
4. *Entropía nula para variables deterministas*: $H(X)=0 \Longleftrightarrow p(x_0)=1$ para algún $x_0$.<br>
   Esto quiere decir que si el resultado es completamente predecible, no hay incertidumbre ni información promedio.
5. *Continuidad*: La entropía es una función continua de la distribución $p(x)$.<br>
   Pequeñas variaciones en las probabilidades producen pequeñas variaciones en la entropía.
6. *Invarianza ante re-etiquetado*: La entropía depende únicamente de la distribución de probabilidad, no de las etiquetas de los valores.<br>
   Renombrar los estados de $X$ no altera la incertidumbre.
7. *Aditividad para variables independientes*: Si $X$ e $Y$ son independientes, $H(X,Y)=H(X)+H(Y)$.<br>
   Esto es, la incertidumbre total de dos fuentes independientes es la suma de sus incertidumbres individuales.
8. *Subaditividad*: Para variables arbitrarias $X,Y$, tenemos que $H(X,Y)\leq H(X)+H(Y)$, con igualdad si y solo si $X$ e $Y$ son independientes.
   <br>
    La dependencia estadística reduce la incertidumbre conjunta.
9. *Monotonía con respecto al refinamiento*: Si $Y$ es una función de $X$, entonces $H(Y)\leq H(X)$.<br>
    Esta propiedad indica que procesar o agrupar información no puede aumentar la incertidumbre.
10. *Concavidad*: La entropía es una función cóncava de la distribución de probabilidad.<br>
   Esto significa que mezclar distribuciones aumenta la incertidumbre promedio.

La **redundancia de X** se define matemáticamente como $1-\dfrac{H(X)}{\log{|\mathcal{A}_X|}}$ y mide la diferencia fraccional entre $H(X)$ y su 
máximo valor posible, $\log{|\mathcal{A}_X|}$.   

Para mayor claridad, trabajemos el siguiente ejemplo. 

El sistema de alertas metereológicas de una cuidad utiliza 4 tipos de alertas codificadas en colores:

1. **VERDE** (Condiciones normales)
2. **AMARILLO** (Precaución)
3. **NARANJA** (Alerta)
4. **ROJO** (Emergencia)

Según datos históricos de 1000 días, las probabilidades de cada alerta son:

|Alerta|Días|Probabilidad| 
|-----|-------|------| 
|Verde|650|p(v)=0.65| 
|Amarillo|250|p(a)=0.25|
|Naranja|80|p(n)=0.08|
|Rojo|20|p(r)=0.02|

*observación*: El sistema transmite las alertas usando un **código binario**. Calcular la entropía $H(X)$ del sistema de alertas (en bits).

Para calcular la **entropía** hacemos uso de su fórmula dada anteriormente por $$H(X)\equiv\sum_{x\in{\mathcal{A}_X}}p(x)\log{\frac{1}{p(x)}}$$, y 
reemplazamos cada probabilidad en la sumatoria, esto es:

$H(X)=p(v)\log{\dfrac{1}{p(v)}}+p(a)\log{\dfrac{1}{p(a)}}+p(n)\log{\dfrac{1}{p(n)}}+p(r)\log{\dfrac{1}{p(r)}}$<br>
$H(X)=0.65*\log_{2}{\dfrac{1}{0.65}}+0.25*\log_{2}{\dfrac{1}{0.25}}+0.08*\log_{2}{\dfrac{1}{0.08}}+0.02*\log_{2}{\dfrac{1}{0.02}}$<br>
$H(X)=1.3084\ bits$

El valor obtenido representa la incertidumbre promedio del sistema de alertas, dado que Verde es altamente probable, la entropía es menor que la 
entropía máxima posible para 4 símbolos, esto es

$$H_{máx}=\log_{2}{(4)}=2\ bits.$$

Este resultado indica que el sistema no es equiprobable, y por tanto puede codificarse eficientemente usando un código binario optimizado.

*observación*: Un **sistema equiprobable** representa una situación de máxima incertidumbre antes de observar el símbolo emitido.

Retornando a la teoria de la información, la **entropía conjunta** supone que en vez de tener una variable aleatoria $X$, existe otra variable $Y$ 
independientes entre sí, es decir el conocimiento de una (por ejemplo, $Y$) entrega información sobre la otra (por ejemplo, $X$). Desde el punto 
de vista de la entropía de la información podemos decir que la información de $Y$ disminuirá la incertidumbre de $X$.

$$H(X,Y)=\sum_{xy\in{\mathcal{A}_X}{\mathcal{A}_Y}}p(x,y)\log\dfrac{1}{p(x,y)}$$

Algunas propiedades fundamentales de la entropía conjunta son:

1. *No Negatividad*: $H(X,Y)\geq0$. La incertidumbre total asociada al par $(X,Y)$ no puede ser negativa.
2. *Cota inferior por las entropías marginales*: $H(X,Y)\geq máx\{H(X),H(Y)\}.$ Conocer simultáneamente dos variables no puede generar menos
   incertidumbre que conocer solo una de ellas.
3. *Cota superior* (subaditividad): $H(X,Y)\leq H(X)+H(Y)$, con igualdad si y solo si $X$ e $Y$ son independientes. Esto es, la dependencia
   estadística reduce la incertidumbre total conjunta.
4. *Independencia*: $X\perp Y\quad \Longleftrightarrow \quad H(X,Y)=H(X)+H(Y)$. Si las fuentes no comparten información, la incertidumbre conjunta
   es simplemente la suma de las incertidumbres individuales.
5. *Simetría*: $H(X,Y)=H(Y,X)$. El orden de las variables no afecta la incertidumbre conjunta.
6. *Entropía conjunta y determinismo*: $Y=f(X)$, entonces $H(X,Y)=H(X)$. Si una variable está completamente determinada por la otra, no se agrega
   incertidumbre adicional al considerar el par.
7. *Monotonía con respecto a la inclusión de variables*: $H(X,Y,Z)\leq H(X,Y)$. Agregar variables nunca reduce la incertidumbre total del sistema.
8. *Continuidad*: La entropía conjunta es continua respecto de la distribución conjunta $p(x,y)$. Pequeños cambios en la ley de probabilidad
   producen pequeños cambios en la incertidumbre.
9. *Concavidad*: La entropía conjunta es una función cóncava de la distribución conjunta $p(x,y)$. Mezclar distribuciones conjuntas incrementa la
    incertidumbre promedio.

Analicemos el siguiente ejemplo, considérese una fuente de información que emite **dos símbolos binarios** $X$ e $Y$, que pueden tomar los valores

$$X\in\{0,1\}, \qquad Y\in\{0,1\}$$.

La distribución de probabilidad conjunta está dada por la siguiente tabla:

|X  | Y |  P(X,Y)|
|------|------|------------|
|0 |	0	| 0.40|
|0 |	1	|0.10|
|1 |	0	|0.20|
|1 |	1	|0.30|

Es inmediato verificar que:

$$\sum_{xy}P(X,Y)=1.$$

Calculemos la entropía conjunta $H(X,Y)$ y analicemos su significado.

Aplicamos directamente la definición término a término.

$H(X,Y)=0.40*\log_2\left(\dfrac{1}{0.40}\right)+0.10*\log_2\left(\dfrac{1}{0.10}\right)+0.20*\log_2\left(\dfrac{1}
{0.20}\right)+0.30*\log_2\left(\dfrac{1}{0.30}\right)$

$H(X,Y)\approx (0.529+0.332+0.464+0.521)=1.846\ bits.$

La entropía conjunta $H(X,Y)$ representa el **número promedio de bits necesarios para codificar sin pérdida** las realizaciones conjuntas de 
$X$ e $Y$. En este ejemplo, observar simultáneamente $X$ e $Y$ genera, en promedio, 1.85 bits de información por símbolo conjunto. Este valor es 
menor que el máximo para 2 bits, lo que indica cierta estructura o dependencia estadística entre $X$ e $Y$.

Además, como la entropía conjunta **cuantifica la incertidumbre global del sistema completo**, lo que incluye la incertidumbre de cada variable y 
las correlaciones o dependencias entre ellas, si $X$ e $Y$ fueran completamente independientes y equiprobables, la entropía conjunta sería máxima.
El valor obtenido sugiere que conocer parcialmente uno de los símbolos reduce la incertidumbre del otro.

Antes de seguir avanzando con los otros tipos de entropía, debemos mencionar la **desigualdad de Gibbs'**.

Se define la **entropía relativa** o **divergencia de Kullback-Leibler** entre dos distribuciones de probabilidad $P(x)$ y $Q(x)$ las cuales se 
definen sobre el mismo conjunto $\mathcal{A}_X$, esto es [4],

$$D_{KL}(P||Q))=\sum_{x\in\mathcal{A}_X}P(x)\log{\frac{P(x)}{Q(x)}}.$$

La entropía relativa satisface la desigualdad de Gibbs'

$$D_{KL}(P||Q))\geq 0$$

con igualdad solo si $P=Q$. Nótese que en general la entropía relativa no es simétrica bajo el intercambio de las distribuciones $P$ y $Q$: en 
general $D_{KL}(P||Q))\neq D_{KL}(Q||P))$ , asi $D_{KL}$, aunque a veces también llamada "distancia $KL$", no es estictamente una distancia. La 
entropía relativa es importante en el reconocimiento de patrones y redes neuronales, así como también en la teoría de la información. 


Volviendo a la teoría de la información, la **entropía condicional** cuantifica la cantidad de información necesaria para describir el resultado 
de una variable aleatoria $X$ dado que el valor de otra variable aleatoria $Y$ es conocido. Aquí, la información se mide en shannons, nats, o 
hartleys. 

Sea $(X,Y)$ un par de variables aleatorias discretas con distribución conjunta $p(x,y)$, la entropía condicional de $X$ dado $Y$ es el promedio, 
sobre $y$, de la entropia condicional de $X$ dado $y$, esto es,

$$H(X|Y)\equiv\sum_{y\in{\mathcal{A}}_Y} p(y)\left[\sum_{x\in{\mathcal{A}}_X} p(x|y)\log{\frac{1}{p(x|y)}}\right]$$
$$=\sum_{xy\in{\mathcal{A}}_X{\mathcal{A}}_Y}p(x,y)\log{\frac{1}{p(x|y)}}.$$

En teoría de la información, el **Teorema de Bayes** no se interpreta únicamente como una regla de actualización probabilística, sino como un 
principio que gobierna la **redistribución de la incertidumbre** entre variables aleatorias. Esta redistribución se expresa naturalmente mediante 
la **entropía conjunta**, la **entropía marginal** y la **entropía condicional**.

Sean $X$ y $Y$ variables aleatorias discretas. El Teorema de Bayes se escribe como

$$p(x|y)=\frac{p(y|x)p(x)}{p(y)}.$$

Multiplicando ambos lados por $p(y)$, se obtiene 

$$p(x,y)=p(x|y)p(y)=p(y|x)p(x).$$

Esta igualdad muestra que el Teorema de Bayes no altera la distribución conjunta: solo ofrece dos factorizaciones equivalentes de $p(x,y)$.

Volviendo a la entropía condicional, a continuación presentamos sus propiedades más importantes. (En algunas propiedades hemos utilizado el Teorema de Bayes)

1. *No negatividad*: $H(X|Y)\geq0$. La incertidumbre promedio sobre $X$ dado $Y$ nunca puede ser negativa. Aun con información auxiliar, no se
   puede tener “incertidumbre negativa”.
2. *Cota superior*: $H(X|Y)\leq H(X)$. Conocer $Y$ no puede aumentar la incertidumbre promedio sobre $X$. La información adicional solo puede
   reducirla o, en el peor de los casos, dejarla igual.
3. *Independencia*: $X\perp Y\quad \Longleftrightarrow \quad H(X|Y)=H(X)$. Si $Y$ no aporta información sobre $X$, entonces conocer $Y$ es
   irrelevante para predecir $X$.
4. *Determinismo condicional*: $H(X|Y)=0 \Longleftrightarrow X=f(Y)$. Si $X$ queda completamente determinado al conocer $Y$, no hay incertidumbre
   residual.
5. *Forma de valor esperado*: $H(X|Y)=\mathbb{E}_Y[H(X|Y=y)]$. La entropía condicional es la incertidumbre promedio de $X$ después de observar
   cada posible valor de $Y$.
6. *Regla de la cadena*: $H(X,Y)=H(Y)+H(X|Y)=H(X)+H(Y|X)$. La incertidumbre total se puede descomponer en una incertidumbre inicial más una
   incertidumbre residual tras observar otra variable.
7. *Subaditividad condicional*: $H(X,Z|Y)\leq H(X|Y)+H(Z|Y)$, con igualdad si $X$ y $Z$ son condicionalmente independientes dado $Y$. Dada cierta
   información común $Y$, la incertidumbre conjunta no supera la suma de las incertidumbres individuales.
8. *Reducción por condicionamiento adicional*: $H(X∣Y,Z)≤H(X∣Y)$. Agregar más información condicionante nunca incrementa la incertidumbre sobre 
    $X$.
9. *Monotonía bajo funciones*: Si $Z=g(Y)$, entonces $H(X|Y)\leq H(X|Z)$. Condicionar con una versión más informativa de $Y$ reduce más la
    incertidumbre que condicionar con una versión degradada.
10. *Simetría indirecta*: En general, $H(X|Y)\neq H(Y|X)$. La incertidumbre residual depende de cuál variable se observa; la relación no es
    simétrica.

Calculemos la entropía condicional $H(X|Y)$ de un sistema de monitoreo clasifica el estado del clima de una ciudad mediante la variable aleatoria

$$X\in\{Soleado,\ Lluvioso\},$$

y registra simultáneamente si se produce o no un congestión vehicular, modelado por la variable

$$Y\in\{Sí,\ No\}.$$

La distribución de probabilidad conjunta $P(X,Y)$ es la siguiente:

|Clima X|Congestión Y|P(X,Y)|
|---|---|------|
|Soleado|	No|	0.50|
|Soleado|	Sí|	0.10|
|Lluvioso|	No|	0.10|
|Lluvioso|	Sí|	0.30|

Para calcular la entropía condicional $H(X|Y)$ debemos primero calcular las probabilidades marginales $P(Y)$, por lo que se obtiene: 

$$P(congestión: Y=No)=0.10+0.30=0.40$$
$$P(congestión: Y=Sí)=0.50+0.10=0.60$$

Ahora, calculamos las probabilidades conjuntas $P(X,Y)$, esto es:

$$P(X:soleado,\ Y=No)=0.5$$
$$P(X:soleado,\ Y=Sí)=0.1$$
$$P(X:lluvioso,\ Y=No)=0.1$$
$$P(X:lluvioso,\ Y=Sí)=0.3$$

Finalmente, aplicando una de las fórmulas para calcular la entropia condicional $H(X|Y)=\sum_{xy}P(X,Y)\log_2{\left(\dfrac{P(Y)}{P(X,Y)}\right)}$ obtenemos:

$H(X|Y)=0.5*\log_{2}{\left(\dfrac{0.6}{0.5}\right)}+0.1*\log_{2}{\left(\dfrac{0.4}{0.1}\right)}+0.1*\log_{2}{\left(\dfrac{0.6}{0.1}\right)}+0.3*\log_{2}{\left(\dfrac{0.4}{0.3}\right)}\approx 0.13152+0.2+0.2585+0.12451  $<br>
$H(X|Y)\approx 0.71453\ bits$

De forma análoga, calculamos $H(Y|X)$ y se obtiene $H(Y|X)\approx 0.71453\ bits$. [*Observación: la igualdad en las entropias no es general, sino consecuencia de la simetría numérica de las probabilidades marginales y condicionales.*]

Por lo tanto, conocer si hay congestión $H(X|Y)$ reduce la incertidumbre sobre el clima en aproximadamente 0.71 bits. Del mismo modo, conocer el clima $H(Y|X)$ reduce la incertidumbre sobre la congestión en la misma magnitud. Por lo tanto, ambas variables están correlacionadas, pero ninguna determina completamente a la otra. 

En teoría de la información, la **información mutua** mide la cantidad de información que dos variables aleatorias comparten. Equivalentemente, 
cuantifica la reducción de incertidumbre en una variable al conocer la otra. 

En palabras simples, la información mutua es capaz de responder ¿Cuánto se reduce la incertidumbre sobre $X$ cuando se conoce $Y$?, en particular:

Si conocer $Y$ no aporta nada sobre $X$, la información mutua es cero.
Si conocer $Y$ determina completamente a $X$, la información mutua es máxima.

Para definir la información mutua, sean $X$ y $Y$ variables aleatorias discretas con distribución conjunta $p(x,y)$ y sus distribuciones marginales son $p(x),\ p(y)$. Matemáticamente, nos queda

$$I(X;Y)=\sum_{xy\in{\mathcal{A}}_X{\mathcal{A}}_Y}p(x,y)\log{\frac{p(x,y)}{p(x)p(y)}}$$

Esta expresión compara la distribución conjunta real con la que existiría si $X$ e $Y$ fueran independientes.

A partir de las definiciones de entropía y entropía condicional, se obtienen identidades equivalentes:

$$I(X;Y)=H(X)−H(X|Y)$$
$$I(X;Y)=H(Y)−H(Y|X)$$
$$I(X;Y)=H(X)+H(Y)−H(X,Y)$$

Estas expresiones hacen explícita la relación de la información mutua con la reducción de incertidumbre y con la entropía conjunta.

Las propiedades mas importantes de la información mutua son:

1. *No negatividad*: $I(X;Y)\geq 0$. Nunca se puede perder información al observar otra variable.
2. *Independencia*: $X\perp Y\quad \Longleftrightarrow \quad I(X;Y)=0$. Si no hay dependencia estadística, no hay información compartida.
3. *Simetría*: $I(X;Y)=I(Y;X)$. La información compartida no depende del orden de las variables.
4. *Cota superior*: $I(X;Y)\leq min\{H(X),H(Y)\}$. No se puede aprender más sobre una variable de lo que ella misma es incierta.
5. *Relación con entropía conjunta*: $I(X;Y)=H(X)+H(Y)−H(X,Y)$. La dependencia reduce la entropía conjunta respecto de la suma marginal.
6. *Cadena de información mutua*: $I(X;Y,Z)=I(X;Y)+I(X;Z|Y)$. La información total se puede descomponer en contribuciones sucesivas.
7. *Monotonía*: $I(X;Y,Z)\geq I(X;Y)$. Agregar observaciones no puede reducir la información disponible.
8. *Relación con divergencia de Kullback–Leibler*: $I(X;Y)=D_{KL}(p(x,y)||p(x)p(y))$. La información mutua mide cuán distinta es la distribución
   conjunta de la que existiría bajo independencia.
9. La **información mutua condicional**: $I(X;Y|Z)=H(X|Z)−H(X|Y,Z)$ y mide la información compartida entre $X$ e $Y$ una vez conocido $Z$.

Si trabajamos con el ejemplo anterior, calculemos la información mutua $I(X;Y)$ del sistema de monitoreo. Para esto nos basaremos en los resultados ya obtenidos, esto es, las probabilidadess marginales, las probabilidades conjuntas y la entropía condicional.

Para las probabilidades marginales $P(Y)$ y $P(X)$ tenemos,

$$P(congestión: Y=No)=0.60\qquad P(soleado)=0.6 $$
$$P(congestión: Y=Sí)=0.40\qquad P(lluvioso)=0.4$$

Para las probabilidades conjuntas $P(X,Y)$ nos queda,

$$P(X:soleado,\ Y=No)=0.5\qquad P(X:lluvioso,\ Y=No)=0.1$$
$$P(X:soleado,\ Y=Sí)=0.1\qquad P(X:lluvioso,\ Y=Sí)=0.3$$

Y finalmente, las entropías condicionales son

$$H(X|Y)=H(Y|X)\approx 0.71453\ bits$$

Con esta información, podemos calcular $I(X;Y)$, usando la fórmula dada por $$I(X;Y)=\sum_{xy\in{\mathcal{A}}_X{\mathcal{A}}_Y}p(x,y)\log{\dfrac{p(x,y)}{p(x)p(y)}}.$$

$I(X;Y)=0.1*\log_2{\left(\dfrac{0.1}{0.6*0.4}\right)}+0.5*\log_2{\left(\dfrac{0.5}{0.6*0.6}\right)}+0.3*\log_2{\left(\dfrac{0.3}{0.4*0.4}\right)}+0.1*\log_2{\left(\dfrac{0.1}{0.4*0.6}\right)}\approx -0.1263+0.23697+0.27027-0.1263$<br>
$I(X;Y)\approx 0.25643$

Primero podemos concluir es que el clima y la congestión vehicular no son independientes ya que $I(X;Y)>0$, además conocer el clima reduce la incertidumbre sobre la congestión en aproximadamente 0.26 bits. Sin embargo, la dependencia es parcial, no determinista: aún existe incertidumbre significativa. Finalmente, el sistema de monitoreo obtiene información útil pero limitada al observar una sola de las variables.

Ahora bien, trabajemos otro ejemplo en el cual tenemos mas variables aleatorias y podemos inferir algunas interpretaciones al comparar mas de un valor de información mutua. 

Sea $X$ una variable aleatoria binaria que representa la condición de **vejez** $(X=1:SI\ \text{o}\ X=0:NO)$ y sea $Y$ una variable 
aleatoria que indica la presencia de **hipertensión** $(Y=1:SI\ \text{o}\ Y=0:NO)$. A partir de una muestra observacional, se estudia la relación 
estadística entre ambas variables utilizando herramientas de la teoría de la información, en particular la entropía y la información mutua, con el 
objetivo de cuantificar el grado de dependencia y la reducción de incertidumbre de una variable al conocer la otra. Para ello se presenta la 
siguiente tabla: 

|Condición | caso 1| caso 2 | caso 3| caso 4| caso 5 | caso 6 |
|---|---|---|---|---|---|---|
|Vejez|	No|	Si| Si|Si|Si| Si|
|Hipertensión|	No|	No| Si|No|Si| No|

De forma análoga al ejemplo anterior, calcularemos las entropias $H(X)$ y $H(Y)$, la entropía conjunta $H(X,Y)$, las entropías condicionales $H(X|Y)$ y $H(Y|X)$ y finalmente la información mutua $I(X;Y)$. Estos resultados son:

$H(X)\approx 0.65\ bits,\qquad H(Y)\approx 0.9183\ bits,\qquad H(X,Y)\approx 1.4591\ bits$,<br>
$H(X|Y)\approx 0.5408\ bits,\qquad H(Y|X)\approx 0.8091\ bits,\qquad I(X;Y)\approx 0.1092\ bits$.  

Lo primero que se observa es que la información mutua es positiva, por lo que vejez e hipertensión no son independientes. Sin embargo $I(X;Y)\approx 0.109\ bits$ es pequeña, indicando una dependencia débil en esta muestra. En conclusión, conocer si una persona es vieja reduce solo ligeramente la incertidumbre sobre la hipertensión.

*Observación*: La diferencia entre $H(X|Y)$ y $H(Y|X)$ refleja que **la vejez es más predecible a partir de la hipertensión que a la inversa**, dado el fuerte sesgo de la muestra hacia $X=1$.

Ahora bien, si cambiamos $X$ que es una variable aleaoria binaria que daba la condición de **vejez** por otra variable aleatoria binaria $X$ que da la condición de **sobrepeso** la cual nos entrega la siguiente tabla:

|Condición | caso 1| caso 2 | caso 3| caso 4| caso 5 | caso 6 |
|---|---|---|---|---|---|---|
|Sobrepeso|	No|	No| Si|No|Si| No|
|Hipertensión|	No|	No| Si|No|Si| No|

De forma análoga, calcularemos las entropias $H(X)$ y $H(Y)$, la entropía conjunta $H(X,Y)$, las entropías condicionales $H(X|Y)$ y $H(Y|X)$ y finalmente la información mutua $I(X;Y)$ dando los siguientes resultados:

$H(X)\approx 0.918\ bits,\qquad H(Y)\approx 0.918\ bits,\qquad H(X,Y)\approx 0.918\ bits$,<br>
$H(X|Y)=0\ bits,\qquad H(Y|X)=0\ bits,\qquad I(X;Y)\approx 0.918\ bits$.  

En este caso, existe una **dependencia perfecta** entre sobrepeso e hipertensión en esta muestra. Conocer si una persona tiene sobrepeso **determina completamente** si tiene hipertensión, y viceversa. La entropía condicional nula indica **ausencia total de incertidumbre residual**. Como $I(X;Y)=H(X)=H(Y)$ esto significa que toda la incertidumbre de cada variable es compartida y desde el punto de vista informacional, el sistema se comporta como si ambas variables fueran la misma fuente.

Por lo tanto, al comparar los dos resultados de la información mutua $I(vejez,hipertensión)\approx 0.109\ bits$ y $I(sobrepeso,hipertensión)\approx 0.918\ bits$ se puede concluir que la relación sobrepeso–hipertensión contiene significativamente mas información compartida que la relacion vejez–hipertensión. Esto significa que, en términos de reducción de incertidumbre, el sobrepeso es un predictor informacional mucho más fuerte de la hipertensión que la vejez.
